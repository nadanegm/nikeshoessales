Importing Pandas

```python
import pandas as pd
data = {
    "Shoe": ["Nike Air Force 1", "Nike Air Max", "Nike Dunk Low",
             "Nike Pegasus", "Nike Jordan 1"],
    "Price": [5500, 6200, 4800, 7000, 8500],
    "Size": [42, 43, 41, 44, 42],
    "Rating": [4.5, 4.8, 4.3, 4.7, 4.9]
}
df = pd.DataFrame(data)
print(df)
```

Matplotib

Importing Matplotib

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Shoe": ["Nike Air Force 1", "Nike Air Max", "Nike Dunk Low",
             "Nike Pegasus", "Nike Jordan 1"],
    "Price": [5500, 6200, 4800, 7000, 8500],
    "Size": [42, 43, 41, 44, 42],
    "Rating": [4.5, 4.8, 4.3, 4.7, 4.9]
}

df = pd.DataFrame(data)

print(df)
```

Charts

Bar Chart

```python
plt.figure(figsize=(10,5))

plt.bar(df["Shoe"], df["Price"], label="Shoe Price")

plt.title("Nike Shoes Prices")
plt.xlabel("Shoe Name")
plt.ylabel("Price (EGP)")

plt.xticks(rotation=30)
plt.grid(axis="y")
plt.legend()

plt.show()
```

Scatter Chart

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Price"], df["Rating"], label="Nike Ratings")

plt.title("Nike Shoes Price vs Rating")
plt.xlabel("Price (EGP)")
plt.ylabel("Customer Rating")

plt.grid()
plt.legend()

plt.show()
```
